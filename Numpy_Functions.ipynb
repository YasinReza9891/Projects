```python
arr_2d = np.array([
    [5,10,15],
    [20,25,30],
    [35,40,45]
])

#INDEXING IN MATRICES
print(f"Element 1,1 -> {arr_2d[0][0]}")
arr_2d[0,0]#this also gives the same output

#SLICING in matrices
print(f"arr_2d[:2,1:] -> {arr_2d[:2,1:]} \n") #Cols 0 TILL (Not including) 2, Rows 1 ONWARDS
print(f"arr_2d[:2] ->{arr_2d[:2]}")
```

```python
arr = np.arange(1,11)

bool_arr = arr>5

bool_arr#CREATES A FILTER THAT CAN BE APPLIED on the OG array

arr[bool_arr]# NEW FILTERED ARRAY

arr[arr>5]#SAME RESULT
```

```python
arr = np.arange(0,10)

# print(arr+arr)
# print(arr-arr)
# print(arr*arr)
# print(arr/arr) #SHOWS runtime Warning not error, if your array contains 0
# print(1/arr) #same
# print(arr**3)


print(f"np.sqrt(arr)  -> {np.sqrt(arr)}\n\n")
print(f"np.exp(arr)  -> {np.exp(arr)}\n\n")
print(f"np.max(arr)  -> {np.max(arr)}\n\n")
print(f"np.min(arr)  -> {np.min(arr)}\n\n")
print(f"np.log(arr)  -> {np.log(arr)}\n\n")
print(f"np.sin(arr)  -> {np.sin(arr)}\n\n")
```
